```python
from datasets import load_dataset

# ============================================================
# Load English SFT dataset
# ============================================================

dataset = load_dataset(
    "Open-Orca/SlimOrca",
    split="train"
)

# First successful test: 5,000 samples
dataset = dataset.shuffle(seed=42).select(range(5000))

df = dataset.to_pandas()

# ============================================================
# Convert to OpenAI-style messages
# ============================================================

contents = []

for _, row in df.iterrows():

    messages = []

    for message in row["conversations"]:

        role_map = {
            "system": "system",
            "human": "user",
            "gpt": "assistant",
        }

        role = role_map.get(message["from"])

        if role is not None:
            messages.append({
                "role": role,
                "content": message["value"],
            })

    contents.append(messages)

df["messages"] = contents

# ============================================================
# Save JSONL
# ============================================================

df[["messages"]].to_json(
    "train.jsonl",
    orient="records",
    lines=True,
    force_ascii=False,
)

print(f"Saved {len(df)} training examples")
print("Output: train.jsonl")
```
